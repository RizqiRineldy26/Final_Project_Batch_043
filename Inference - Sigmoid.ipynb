```python
# Import libraries

import gdown
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
# link_id = '1Af0XnDD7AigbEWXP11EPlRMcu6sglHRU' # MODEL2
# link_id = '19Y3ABz0UBd2MCSNis-hqBBonP5p_feA7' # MODEL3
link_id = '1RF6--tsLehbyThjtzER_IMLXRn6FJ6pf'
url = f'https://drive.google.com/uc?id={link_id}'
output_path = 'model_2.keras'

gdown.download(url, output_path, quiet=False)
model = tf.keras.models.load_model(output_path)
```

```python
file_1 = '1pQuaroakmEEgXawC5k2441nYlhePL6k1'
file_2 = '1vvIVUDGfL8Lw4-iZ8uc1xQjARb3WK-Zk'
file_3 = '1NgTDHa2mggFVdPK-qafxG-3splt4H9VV'
file_4 = '1l-Fc-faQPXZr2fB9CedrDhXq72dx9Vu0'
file_5 = '1NBOwiTgUjj5jvmCrXJkWvip0JTIinOrs'
file_6 = '1FS9ONd87T5XfXi1M6IKw1hgARWIN0uJA'
file_7 = '1KxyhZG7_nhmEXBk5QFROTwzXmE9VykMj'
file_8 = '1VlEGUxqGwDv04mU2EI-spX7DX5jXpURg'
file_9 = '1dh_t_6Umy_OOZ_z0p9OX_xP4JOK8Gj3X'

def inference(file_name, file_id, img_height=150, img_width=150):
  url = f'https://drive.google.com/uc?id={file_id}'
  output_path = f'downloaded_image_{file_name}.jpg'
  gdown.download(url, output_path, quiet=False)

  img = tf.keras.utils.load_img(output_path, target_size=(img_height, img_width))
  img_array = tf.keras.utils.img_to_array(img)
  img_array = tf.expand_dims(img_array,0)

  pred_prob = model.predict(img_array)
  pred_class = np.argmax(pred_prob[0])
  class_names = ['acne', 'blackheads', 'dark_spots', 'pores', 'wrinkles']
  pred_class_name = class_names[pred_class]
  pct_prob = [f'{x * 100:.2f}%' for x in pred_prob[0]]
  inf_df = pd.DataFrame({
      'breed':class_names,
      'prediction': pct_prob
  })

  return img, inf_df, pred_class_name
```

```python
img, inf_df, pred_class_name = inference('file_1', file_1)
img2, inf_df2, pred_class_name2 = inference('file_2', file_2)
img3, inf_df3, pred_class_name3 = inference('file_3', file_3)
img4, inf_df4, pred_class_name4 = inference('file_4', file_4)
img5, inf_df5, pred_class_name5 = inference('file_5', file_5)
img6, inf_df6, pred_class_name6 = inference('file_6', file_6)
img7, inf_df7, pred_class_name7 = inference('file_7', file_7)
img8, inf_df8, pred_class_name8 = inference('file_8', file_8)
img9, inf_df9, pred_class_name9 = inference('file_9', file_9)
```
