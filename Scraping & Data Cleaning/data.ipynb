## Loading data

Membuat data base untuk setiap table

```python
df_acne = pd.read_csv("Skincare Dataset/Anti-Acne Product.csv")
df_comedo = pd.read_csv("Skincare Dataset/Comedo Product.csv")
df_darkspots = pd.read_csv("Skincare Dataset/Dark-Spots Product.csv")
df_pores = pd.read_csv("Skincare Dataset/pores_data.csv")
df_wrinkle = pd.read_csv("Skincare Dataset/wrinkle.csv")
```

### Final Cleaning

Kolom Problem

```python
df_clean["problem"].str.strip().str.lower()
display(df_clean["problem"])
```

#### Membersihkan Kolom Product Name

```python
df_clean["product_name"] = (
df_clean["product_name"].str.replace(r"^\s*\d+\.?\d*%\s*\|?\s*", "", regex=True)
.str.replace(r"\s+", " ", regex=True)
.str.strip()
)
df_clean = df_clean[~df_clean["product_name"].str.contains("Next page >>", case=False, na=False)]
df_clean = df_clean[~df_clean["product_name"].str.contains("Show all >>", case=False, na=False)]

df_clean = df_clean.reset_index(drop=True)

display(df_clean["product_name"])
```

Terlihat nama produk menjadi lebih rapi dan mudah di baca

Membersihkan kolom `ingredient`

```python
def clean_ingredients(text):
    ingredients = text.split(",")

    cleaned = []

    for i in ingredients:
        i = i.strip()
        cleaned.append(i)

    cleaned =  list(dict.fromkeys(cleaned))

    cleaned = cleaned[:10]

    return ", ".join(cleaned)

df_clean["ingredient"] = df["ingredient"].apply(clean_ingredients)
df_clean
```

Karena jumlah ingredient setiap product bervariasi, maka untuk kerapian data setiap produk dibatasi hanya menggunakkan 10 ingredient pertama.

Pembuatan fungsi upload

```python
def upload_data(df_clean):
    query = text("""
        INSERT INTO skincare_recommendation
        (problem, ingredient, product_name, product_link)
        VALUES
        (:problem, :ingredient, :product_name, :product_link)
        ON CONFLICT (problem, ingredient, product_name, product_link)
        DO NOTHING
    """)

    data = df_clean[
        ["problem", "ingredient", "product_name", "product_link"]
    ].to_dict(orient="records")
    
    with engine.begin() as conn:
        result = conn.execute(query, data)

print("Data baru yang masuk:", result.rowcount)
```

Fungsi upload_data() digunakan untuk memasukkan data ke tabel skincare_recommendation.

Upload Data

```python
upload_data(df_clean)
```
