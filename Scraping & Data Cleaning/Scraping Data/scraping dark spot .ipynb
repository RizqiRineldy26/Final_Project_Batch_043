## Import Libraries

```python
import pandas as pd 
from selenium import webdriver
from bs4 import BeautifulSoup
import time
```

## Load Data

```python
df = pd.read_csv("Skincare Dataset/Dark-Spots Product.csv")
df = df.loc[:, ~df.columns.str.contains("^Unnamed: 0")]

df.head(10)
```

Dari 10 data awal yang ditampilkan terlihat ada 4 kolom yaitu :

- `Problem` : menunjukkan jenis permasalahan kulit data tersebut adalah dark spots.

- `ingredient` : bahan apa saja yang terkandung dalam setiap produk.

- `Product Name` : nama produk yang sesuai dengan permasalah kulit tersebut.

- `Product_link` : berisi link dari produk yang akan menampilkan gambar dan detail produk tersebut.

##### Membuat Driver untuk Chrome Browser

```python
driver = webdriver.Chrome()
```

##### Scraping Data

```python
for index, row in df.iterrows():
    url = row["product link"]

    driver.get(url)
    time.sleep(2)

    soup = BeautifulSoup(driver.page_source, "html.parser")

    ingredient_elements = soup.select("a.ingred-link")

    ingredients = []

    for ing in ingredient_elements:
        ingredient_name = ing.get_text(strip=True)

        if ingredient_name not in ingredients:
            ingredients.append(ingredient_name)

    if len(ingredients) > 0:
        df.loc[index, "ingredient"] = ", ".join(ingredients)

    print("Product:", row["product name"])
    print("Link:", url)
    print("Jumlah ingredient:", len(ingredients))
    print("Ingredient:", ingredients)
    print("-" * 50)
    
driver.quit()
```

Hal tersebut dilakukan untuk menambahkan kolom `ingredient` yang sebelumnya hanya terdapat 1 untuk setiap kelompok produk. 

Oleh karena itu, scraping data dilakukan agar kolom ingredient per product lebih lengkap dan detail.

Pada masalah kulit Darkspots. Setiap produk mempunyai jumlah yang ingredients yang bervariasi. Satu produk mempunyai sekitar 2 - 47 ingredients yang berbeda-beda.


##### Penyimpanan data

```python
df.to_csv("Skincare Dataset/Dark-Spots Product.csv", index=False)

df.head(10)
```
